```python
import os
import requests

API_URL = "https://router.huggingface.co/v1/chat/completions"
headers = {
    "Authorization": f"Bearer {os.environ['HF_TOKEN']}",
}  # Bearer : keyword , 

def query(payload):  # what we are going to access that is payload 
    response = requests.post(API_URL, headers=headers, json=payload) # using to access , router: navigating the model address, API_URL : path / location , headers : directly it is not able to access ..so we need access token...headers mostly we use for to access the tokens ,
    # whenever we are using API the response should be int the form of json , it only acceps json, so json=payload
    return response.json()

response = query({
    "messages": [
        {
            "role": "user",
            "content": "What is the capital of France?"
        }
    ],
    "model": "deepseek-ai/DeepSeek-R1:novita"
})  # response will be in the form of json

print(response["choices"][0]["message"])

# success code: 200 (successfully writtened the result )

# first we should check first reponse 
# if we reach the token limit then it will throw the error 
```
